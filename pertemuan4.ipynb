```python
# 1 menentukan tahun kabisat
tahun = int(input("Masukkan tahun: "))
is_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
if is_kabisat:
  print(f"Tahun {tahun} adalah TAHUN KABISAT.")
else:
  print(f"Tahun {tahun} BUKAN tahun kabisat.")

# 2 program sederhana
nama_barang = input("Masukkan nama barang: ")
harga = float(input("Masukkan harga satuan (Rp): "))
jumlah = int(input("Masukkan jumlah barang: "))

total_kotor = harga * jumlah
diskon = total_kotor * 0.10 #Diskon 10%
total_bayar = total_kotor - diskon

print("\n--- Rincian Pembayaran ---")
print(f"Barang      : {nama_barang}")
print(f"Total Kotor : Rp {total_kotor:,.2f}")
print(f"Diskon 10%  : Rp {diskon:,.2f}")
print(f"Total Bayar : Rp {total_bayar:,.2f}")
```
