```python
# AKTIVITAS 6E

x = [10, 20, 30]
y = x
z = x[:]

y.append(40)
z.append(50)
x.insert(0, 5)

print("x:", x)
print("y:", y)
print("z:", z)
```

Mengapa z berbeda?
Karena y = x hanya membuat nama y merujuk ke list yang sama dengan x, sehingga perubahan pada y juga memengaruhi x. Sedangkan z = x[:] membuat salinan list baru, sehingga perubahan pada x tidak memengaruhi z.

```python
# AKTIVITAS 6F

skor = [80, 65, 90, 75, 70]

total = 0
jumlah_lulus = 0

for nilai in skor:
    total = total + nilai
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
# AKTIVITAS 6G

suhu = [28, 31, 30, 33, 29, 27, 32]

rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")

# Karena dari 7 data suhu, rata-rata bergerak 5 hari hanya dapat dihitung sebanyak 3 kali.
# Setiap perhitungan menggunakan 5 data suhu yang bergeser satu hari, sehingga menghasilkan 3 nilai rata-rata.
```

Pada hari ke-17, mesin hanya beroperasi selama 18 jam, sedangkan hari lainnya 24 jam. Hal ini bisa menyebabkan energi yang dihasilkan lebih sedikit. Untuk memastikannya, perlu diperiksa catatan jam operasi dan waktu mesin berhenti.

Penurunan energi mungkin disebabkan oleh gangguan atau kerusakan pada mesin sehingga kinerjanya tidak maksimal. Data yang perlu diperiksa adalah laporan kerusakan dan catatan perawatan mesin.
3. Pengaruh kondisi lingkungan
Kondisi lingkungan yang berubah bisa memengaruhi kinerja alat dan hasil produksi energi. Untuk memastikannya, perlu diperiksa data suhu lingkungan dan kondisi alat pada hari ke-17.

```python
# AKTIVITAS 6I

suhu = [28, 30, 27, 32, 29, 31, 26]

# Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# Menambahkan dan mengeluarkan data
suhu.append(33)
print("Setelah append:", suhu)
suhu.pop()
print("Setelah pop:", suhu)

# Menampilkan suhu yang sudah diurutkan
print("Suhu terurut:", sorted(suhu))
print("List asli:", suhu)

# Menghitung rata-rata dan suhu tertinggi
rata = sum(suhu) / len(suhu)
tertinggi = max(suhu)
hari_tertinggi = suhu.index(tertinggi) + 1

print("Rata-rata:", round(rata, 2))
print("Suhu tertinggi:", tertinggi, "°C")
print("Suhu tertinggi terjadi pada hari ke-", hari_tertinggi)

# Membuat list suhu di atas rata-rata
di_atas_rata = []
for s in suhu:
    if s > rata:
        di_atas_rata.append(s)

print("Suhu di atas rata-rata:", di_atas_rata)
```

```python
# AKTIVITAS 6J

nilai = [
    [80, 70, 90],     # baris 0: Ana
    [65, 75, 60],     # baris 1: Budi
    [90, 85, 95],     # baris 2: Cici
]

print(nilai[1])
print(nilai[1][2])
print("baris:", len(nilai), "kolom:", len(nilai[0]))

# Total untuk setiap mahasiswa
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Skor ujian 1 semua mahasiswa
ujian1 = []
for baris in nilai:
    ujian1.append(baris[0])

print("Skor ujian 1:", ujian1)

# Skor ujian 2 milik Cici
print("Nilai ujian 2 Cici:", nilai[2][1])

# Jumlah ujian (kolom)
print("Jumlah ujian:", len(nilai[0]))
```

```python
# Lima latihan kecil (data kecil)

skor = [80, 65, 90, 75, 70, 85]

# 1. Mencetak skor pertama dan terakhir
print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

# 2. Menghitung rata-rata
rata = sum(skor) / len(skor)
print("Rata-rata:", rata)

# 3. Menghitung skor di atas rata-rata
jumlah = 0
for nilai in skor:
    if nilai > rata:
        jumlah += 1

print("Jumlah skor di atas rata-rata:", jumlah)

# 4. Mencari posisi skor terendah
terendah = min(skor)
posisi = skor.index(terendah) + 1
print("Mahasiswa dengan skor terendah:", posisi)

# 5. Membuat list baru dengan setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)

print("Skor setelah ditambah 5:", skor_baru)
```

```python
# Laporan produksi bulanan (portofolio)

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

target = 6000

# 1. Total dan rata-rata energi
total = sum(energi)
rata = total / len(energi)

print("Total energi:", total, "MWh")
print("Rata-rata energi:", round(rata, 2), "MWh")

# 2. Tiga hari tertinggi dan terendah
tinggi = sorted(energi, reverse=True)[:3]
rendah = sorted(energi)[:3]

print("\n3 hari tertinggi:")
for nilai in tinggi:
    hari = energi.index(nilai) + 1
    print("Hari", hari, ":", nilai, "MWh")

print("\n3 hari terendah:")
for nilai in rendah:
    hari = energi.index(nilai) + 1
    print("Hari", hari, ":", nilai, "MWh")

# 3. Jumlah hari di bawah target
di_bawah = []
for i in range(len(energi)):
    if energi[i] < target:
        di_bawah.append(i + 1)

print("\nJumlah hari di bawah target:", len(di_bawah))
print("Hari:", di_bawah)

# 4. Rata-rata bergerak 7 hari di bawah 97% rata-rata bulanan
batas = rata * 0.97

print("\nHari dengan rata-rata bergerak 7 hari di bawah 97% rata-rata bulanan:")
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7

    if rata_7 < batas:
        print("Hari", i + 1, ":", round(rata_7, 2), "MWh")

# 5. Total energi per minggu
print("\nTotal energi per minggu:")

for i in range(0, len(energi), 7):
    minggu = energi[i : i + 7]
    nomor_minggu = (i // 7) + 1
    print("Minggu", nomor_minggu, ":", sum(minggu), "MWh")
```
