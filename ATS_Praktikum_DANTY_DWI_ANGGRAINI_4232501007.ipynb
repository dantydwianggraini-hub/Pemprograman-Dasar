```python
# SOAL : 5 KATUP PENGAMAN
# NAMA : DANTY DWI ANGGRAINI
# NIM  : 4232501007
# TANGGAL : 9 November 2026
# MATA KULIAH : PEMPROGRAMAN DASAR

# Input tekanan uap
tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

# Status manual override
status_manual_override = False

# Mentukan status katup
if tekanan_uap > 150 and status_manual_override == False:
  print("Katup Terbuka")
elif status_manual_override == True:
  print("Katup Mode Manual")
else:
  print("Katup Tertutup")

# Penjelasan
# float(input()) digunakan untuk membaca tekanan uap dalam bentuk bilangan desimal.
# if mengecek apakah tekanan lebih dari 150 psi dan manual override bernilai False.
# elif mengecek apakah manual override bernilai True.
# else dijalankan jika kedua kondisi sebelumnya tidak terpenuhi.
```

```python
# SOAL : 7 PEMANAS MENCAPAI TARGET
# NAMA : DANTY DWI ANGGRAINI
# NIM  : 4232501007
# TANGGAL : 9 November 2026
# MATA KULIAH : PEMPROGRAMAN DASAR

suhu_saat_ini = 60
while suhu_saat_ini < 85:
  suhu_saat_ini += 5
  print("Suhu saat ini:", suhu_saat_ini, "C")

if suhu_saat_ini == 80:
  print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")

# Penjelasan
# setiap perulangan, suhu dinaikkan sebesar 5 °C.
# operator += berarti menambahkan nilai ke variabel yang sudah ada.
# misalnya, suhu 60 menjadi 65, kemudian 70, 75, 80, dan 85.
# perintah print() digunakan untuk menampilkan suhu ke layar.
# huruf f pada f-string digunakan agar nilai variabel bisa dimasukkan langsung ke dalam teks.
# suhu ditampilkan setelah dinaikkan sebesar 5 °C.
```

```python
# SOAL : 9 Pembacaan Vibrasi dengan Sensor Rusak
# NAMA : DANTY DWI ANGGRAINI
# NIM  : 4232501007
# TANGGAL : 9 November 2026
# MATA KULIAH : PEMPROGRAMAN DASAR

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

data_tidak_valid = 1

for i in range(len(data)):
  unit = i + 1

  try:
        nilai = float(data[i])

  except ValueError:
        print(f"Unit {unit}: data tidak valid")
        data_tidak_valid += 1
        continue

  if nilai > 2.0:
    print(f"Warning: Unit {unit} vibrasi tinggi!")

  if nilai > 3.0:
    print("KRITIS: pemeriksaan dihentikan")
    break

  print(f"Jumlah data tidak valid: {data_tidak_valid}")

# Penjelasan
# Program ini menggunakan for untuk mengecek data setiap unit
# try-except untuk menangani data yang tidak valid
# continue untuk melewati data error
# dan break untuk menghentikan pengecekan jika vibrasi melebihi 3.0.
# diakhir program, ditampilkan jumlah data yang tidak valid
```
