1. Penjelasan ekspresi 10 + 5 x 2 :
Ekspresi 10 + 5 x 2 menghasilkan nilai 20 (bukan 30) karena Python menerapkan aturan urutan prioritas operator (operator precedence).
Dalam prioritas operator Python :
a.  Operator perkalian(*) memiliki prioritas/preseden yang lebih tinggi daripada operator penjumlahan(+).
b. Oleh karena itu, Python akan mengevaluasi dan menyelesaikan operasi perkalian terlebih  dahulu : 5 x 2 = 10
c. Setelah itu, hasil perkalian ditambahkan ke angka 10 : 10 + 10 = 20
Jika kita ingin hasilnya bernilai 30, kita harus menggunakan tanda kurung () yang memiliki tingkat prioritas paling tinggi: (10+ 5) * 2
Sumber referensi: Dokumentasi resmi python (Puthon Documentation):Section 6.17. Opperator precedence (https://docs.python.org/3/reference/expressions.html#operator-precedence)

2. Perbedan mendasar
- OPerator Relasional(.>,<,==,>=,<=,!=): Digunakan untuk membandingkan dua nilai atau data. Hasil dari operasi relasional adalah nilai boolean (True atau False).
- Operator Logika (and, or, not): Digunakan untuk menggabungkan atau memanipulasi beberapa nilai kondisi boolean. Operator ini menentukan logika keseluruhan dari beberapa syarat yang dievaluasi.

Contoh situasi nyata(penggunaan bersamaan):
Skenario: Sistem Verifikasi Kelayakan Kredit Pemilikan Rumah(KPR) Bank.
Untuk menyetujui pengajuan KPR, bank menetapkan dua syarat sekaligus:
1. Usia pemohon minimal21 tahun (usia >= 21)
2. Penghasilan bulanan minimal Rp5.000.000 (penghasilan >= 5000000)
Penjelasan:
- usia >= 21 dan penghasilan >= 5000000 menggunakan operator relasional untuk mengevaluasi dat individu.
- Operator logika and digunakan untuk memastikan kedua syarat tersebut harus bernilai True bersamaan agar keputusan kredit disetujui. Jika salah satu syarat tidak terpenuhi, maka hasilnya adalah False.

3. Contoh kasus bug akibat salah memahi operator precedence
- Ringkasan kasus:
Pada pembuatan fitur kalkulator rata-rata atau diskon(seperti di forum Stack Overflow), seorang pengembang ingin menghitung rata-rata dari dua angka dengan menuliskan kode rata_rata = a + b / 2. Karena operator pembagian (/) memiliki prioritas lebih tinggi dibanding penjumlahan (+), Python mengevaluasi b / 2 terlebih dahulu sebelum menambahkan nilainya ke a. Hal ini menyebabkan kalkuasi salah karena nilai a tidak ikut dibagi dua. Bug ini tidak menghasilkan error program (syntax error), melainkan kesalahan logika (logical error) yang membuat hasil perhitungan keuangan atau statistik menjadi salah.
- Cara perbaikan kode:
Gunakan tanda kurung () untuk mengelompokkan operasi penjumlahan agar dijalankan terlebih dahulu sebelum pembagian.
# Kode salah:
rata_rata = a + b / 2
# Kode Benar:
rata_rata = (a + b) / 2
